```python
## Basic syntax rules in pyhton
## Case sensitivity: Python is case-sensitive, meaning that variables, function names, and other identifiers must be used consistently with the same capitalization. For example, `myVariable` and `myvariable` would be considered two different identifiers.

name='Ahmed'
Name='Ali'

print(name)  # Output: Ahmed
print(Name)  # Output: Ali
```

### Indentation: 
Python uses indentation to define blocks of code. Consistent indentation is crucial for the correct execution of code. For example, a function definition and its body must be indented consistently.

```python
# Line Continuation use the \ to contine the code in the next line
total = 1 + 2 + 3 + \
        4 + 5
print(total)  # Output: 15
```

```python
# multible statements on a single line using semicolons
x=5; y=10; z=15; sum=x+y+z; 
print(sum)  # Output: 30
```
