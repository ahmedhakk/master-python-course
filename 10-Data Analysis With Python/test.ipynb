```python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Laptop", "Phone"],
    "Sales": [1000, 500, 1500, 700],
    "Region": ["East", "West", "West", "East"],
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# 1. Display the first two rows of the DataFrame.
df.head(2)
```

```python
# 2. Select only the Product and Sales columns.
df[["Product", "Sales"]]
```

```python
# 3. Find all rows where Sales is greater than 600.
df[df["Sales"] > 600]
```

```python
# 4. Calculate the total Sales for each Product.
total_sales_by_product = df.groupby("Product")["Sales"].sum()
total_sales_by_product
```

```python
# 5. Create a bar chart showing total Sales by Product.
import matplotlib.pyplot as plt

categories = ["Laptop", "Phone"]
values = [total_sales_by_product["Laptop"], total_sales_by_product["Phone"]]

plt.bar(categories, values, color="purple")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Bar Plot")
plt.show()
```

```python
total_sales_by_product.plot(kind="bar")

plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.title("Total Sales by Product")
plt.show()
```
