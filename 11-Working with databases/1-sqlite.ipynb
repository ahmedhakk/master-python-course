### SQL And SQLite
SQL (Structured Query Language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. In this lesson, we will cover the basics of SQL and SQLite, including creating databases, tables, and performing various SQL operations.

```python
import sqlite3
```

```python
# Connect to an SQLite database
connection = sqlite3.connect("example.db")
connection
```

```python
cursor = connection.cursor()
```

```python
# Create a Table into the DB
cursor.execute("""
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department Text
    )
""")

# commit the changes
connection.commit()
```

```python
cursor.execute("""
Select * from employees
""")
```

```python
## Insert the data in sqlite table
cursor.execute("""
INSERT INTO employees(name, age, department)
                VALUES('Ahmed', 23, 'Data Scientist')
""")

cursor.execute("""
INSERT INTO employees(name, age, department)
                VALUES('Bob', 25, 'Software Engineer')
""")

cursor.execute("""
INSERT INTO employees(name, age, department)
                VALUES('Krish', 34, 'Data Scientist Instructor')
""")

# Commit the changes in DB
connection.commit()
```

```python
# Query the date from the table
cursor.execute("SELECT * from employees")
rows = cursor.fetchall()

# print the queried data
for row in rows:
    print(row)
```

```python
## Update the data in the table
cursor.execute("""
UPDATE employees
SET age = 24
WHERE name="Ahmed"
""")
connection.commit()
```

```python
# Query the date from the table
cursor.execute("SELECT * from employees")
rows = cursor.fetchall()

# print the queried data
for row in rows:
    print(row)
```

```python
## Delete the data from the table
cursor.execute("""
DELETE FROM employees
            WHERE name = "Bob"
""")
connection.commit()
```

```python
# Query the date from the table
cursor.execute("SELECT * from employees")
rows = cursor.fetchall()

# print the queried data
for row in rows:
    print(row)
```

```python
## Working Wwith Sales Data
# Connect to an SQLite database
connection = sqlite3.connect("sales_data.db")
cursor = connection.cursor()

# Create a table for sales data
# Delete the incorrectly created table
cursor.execute("DROP TABLE IF EXISTS sales")

# Recreate the table correctly
cursor.execute("""
CREATE TABLE sales (
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
)
""")

connection.commit()

# Insert data into the sales table
sales_data = [
    ("2023-01-01", "Product1", 100, "North"),
    ("2023-01-02", "Product2", 200, "South"),
    ("2023-01-03", "Product1", 150, "East"),
    ("2023-01-04", "Product3", 250, "West"),
    ("2023-01-05", "Product2", 300, "North"),
]
```

```python
cursor.execute("PRAGMA table_info(sales)")
print(cursor.fetchall())
```

```python
# Inserting bulk of data

cursor.executemany(
    """
Insert into sales(date,product,sales,region)
                   values(?,?,?,?)
""",
    sales_data,
)

connection.commit()
```

```python
# Query the date from the table
cursor.execute("SELECT * from sales")
rows = cursor.fetchall()

# print the queried data
for row in rows:
    print(row)
```

```python
# close the connection
connection.close()
```
