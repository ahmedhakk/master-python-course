#### Decorators
Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

### Closures

```python
def main_welcome():
    msg = "welcome"

    def sub_welcome_method():
        print("welcome to advance python course")
        print(msg)

        print("Please learn this concepts")

    return sub_welcome_method()
```

```python
main_welcome()
main_welcome()
main_welcome()
```

```python
def main_welcome(func, lst):
    def sub_welcome_method():
        print("welcome to advance python course")
        print(func(lst))
        print("Please learn this concepts")

    return sub_welcome_method()
```

```python
main_welcome(len, [1, 2, 3, 4, 5, 6, 7])
```

### Decorators

```python
def main_welcome(func):
    def sub_welcome_method():
        print("welcome to advance python course")
        func()
        print("Please learn this concepts")

    return sub_welcome_method()
```

```python
def course_introduction():
    print("this is an advanced python course")
```

```python
main_welcome(course_introduction)
```

when ever we write @function and then define a function under it => it will be pass to the @function as a parameter

```python
@main_welcome
def course_introduction():
    print("this is an advanced python course")
```

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")

    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
# Decorators with arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)

        return wrapper

    return decorator
```

```python
@repeat(3)
def say_hello():
    print("Hello")
```

```python
say_hello()
```

### Docs

```python
# Without nonlocal count, Python would treat count as a local variable inside increment() when assigning to it, causing an UnboundLocalError.
# nonlocal allows the inner function to modify the variable from its enclosing function.

def counter():
    count = 0

    def increment():
        nonlocal count
        count += 1
        return count

    return increment


my_counter = counter()

print(my_counter())  # 1
print(my_counter())  # 2
print(my_counter())  # 3
```
