### Iterators
Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators provide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
my_list = [1, 2, 3, 4, 5, 6]
for i in my_list:
    print(i)
```

```python
print(my_list)
```

```python
## iterator
iterator = iter(my_list)
print(type(iterator))
```

```python
# String iterator
my_string = "Hello"
string_iterator = iter(my_string)

print(next(string_iterator))  # Output: H
print(next(string_iterator))  # Output: e
```

```python
# Creating an iterator
class MyNumbers:
    def __iter__(self):
        self.a = 1
        return self

    def __next__(self):
        if self.a > 5:
            raise StopIteration

        x = self.a  # x = 1
        self.a += 1  # self.a = 2
        return x  # returns 1


my_class = MyNumbers()
myiter = iter(my_class)

print(next(myiter))  # 1
print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
```
