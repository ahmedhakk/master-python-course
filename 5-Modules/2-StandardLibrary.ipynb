```python
## Data Serialization
import json
data = {"name": "Krish", "age": 25}

json_str = json.dumps(data)  # convert data into a json formate
print(json_str)
print(type(json_str))

parsed_data = json.loads(json_str)  # convert json back to dictionary
print(parsed_data)
print(type(parsed_data))
```

```python
# csv
import csv

with open("example.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "age"])
    writer.writerow(["ahmed", 23])

with open("example.csv", mode="r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
import pandas as pd

df = pd.read_csv("example.csv")

print(df)
```
