```python
for i in range(1, 6): # last number is not included, so it will print 1,2,3,4,5
    print(i)
```

```python
# Strings
str = "Ahmed Ali"

for i in str:
    print(i)
```

```python
# While loop
count = 0
while count < 5:
    print(count)
    count += 1
```

```python
count = 0
while count % 2 == 0:
    print(count)
    count += 1
```

```python
# Loop control statements
# break

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# Continue
# The continue statement is used to skip the current iteration of a loop and move on to the next iteration. When the continue statement is encountered, the rest of the code inside the loop for that particular iteration is skipped, and the loop proceeds to the next iteration.

for i in range(10):
    if i % 2 == 0:  # Check if the number is even
        continue
    print(i)
```

```python
# Pass 
# the pass statment is a null operation; it does nothing

for i in range(5):
    if i == 2:
        pass  # Do nothing when i is 2

    print(i)
```

```python
# Nested Loops
for i in range(3):  # Outer loop
    for j in range(2):  # Inner loop
        print(f"Outer loop iteration: {i}, Inner loop iteration: {j}")
```

```python
# Example - calc the sum of n numbers

# while loop
n = 10
sum = 0
count = 1
while count <= n:
    sum += count
    count += 1

print(f"The sum of the first {n} numbers is: {sum}")

# for loop
sum2 = 0
for i in range(n + 1):
    sum2 += i

print(f"The sum of the first {n} numbers is: {sum2}")
```

```python
# Ex: Prime numbers between 1 and 100

for num in range(1, 101):
    if num > 1:  # Prime numbers are greater than 1
        for i in range(2, num):  # Check for factors up to the square root of num
            if (num % i) == 0:
                break
        else:
            print(num)
```
