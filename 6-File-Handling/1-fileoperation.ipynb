```python
### Read a Whole File

with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
# read file line by line
with open("example.txt", "r") as file:
    for row in file:
        print(row.strip())  ### sstrip() removes the newline character
```

```python
## Writing a file (Overwriting) -> remove all content in the file and overwrite it with this new content

with open("example.txt", "w") as file:
    file.write("Hello World\n")
    file.write("this is a new line.")
```

```python
# write a file (without overwriting)
with open("example.txt", "a") as file:  # 'a' append
    file.write("\nAppend operation taking place!\n")
```

```python
### Writing a list of lines to a file
lines = ["First line \n", "Second line \n", "Third line\n"]

with open("example.txt", "a") as file:
    file.writelines(lines)
```

```python
### Binary Files => example.bin

# Writing to a binary file
data = b"\x00\x01\x02\x03\x04"

with open("example.bin", "wb") as file:  # wb => write bite
    file.write(data)
```

```python
with open("example.bin", "rb") as file:  # rb => read bite
    content = file.read()
    print(content)
```

```python
### Read the content from a source text file and write to a destination text file
with open("example.txt", "r") as source_file:
    content = source_file.read()

with open("destination.txt", "w") as destination_file:
    destination_file.write(content)
```

```python
# Read a text file and count the number of lines, words, and characters.
def count_text_file(file_path):
    with open(file_path, "r") as file:
        lines = file.readlines()

        lines_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)

        return lines_count, word_count, char_count


lines, words, chars = count_text_file("example.txt")

print(f"lines: {lines}, words: {words}, chars: {chars}")
```

The w+ mode in Python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exists, its content is truncated (i.e., the file is overwritten).

```python
## Writing and then reading a file
with open("example.txt", "w+") as file:
    file.write("hello world\n")
    file.write("this is a new line\n")

    # Move the file cursor to the beginning
    # if we remove this and then try to read the file we can't see any thing printed! why? - because the cursor is in the end of the file as it was writing so we need to move the file cursor to the top to can read it.
    file.seek(0)

    # read content of the file
    content = file.read()
    print(content)
```
