```python
# accessing dictionary elements
student = {"name": "Krish", "age": 32, "grade": "A"}

print(student)
```

```python
# Accessing dictionary elements
print(student["grade"])

# accessing using get() method
print(student.get("grade"))
print(student.get("last_name"))
print(
    student.get("last_name", "Not in the dictionary")
)  # obj.get(key, default value if None)
```

```python
# Modifing dictionary elements
# Dictionary are mutable,so you can add, update or delete elements

print(student)
```

```python
student["age"] = 33
print(student)
student["address"] = "India"
print(student)
```

```python
del student["grade"]  ## delete key and value pair
print(student)
```

```python
## Dictionary methods
keys = student.keys()
print(keys)
values = student.values()
print(values)

items = student.items()  # get all the keys and values pairs
print(
    items
)  # list of tuples # ([('name', 'Krish'), ('age', 33), ('address', 'India')])
```

```python
# Shallow copy
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "Krish2"
print(student)
print(student_copy)
```

```python
student_copy1 = (
    student.copy()
)  # Shallow copy of the dict - which it refer to a different object in the memory .. so we have 2 seperate values in each variable.

print(student)
print(student_copy1)
```

```python
student["name"] = "Krish"
print(student_copy1)
print(student)
```

```python
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964,
    "nested": {"name": "ahmed", "email": "ahmed@gmail.com"},
}
mydict = thisdict.copy()
print(mydict)

thisdict["nested"]["name"] = "mohamed"
print(thisdict)
print(mydict)
```

```python
import copy

thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964,
    "nested": {"name": "ahmed", "email": "ahmed@gmail.com"},
}

mydict = copy.deepcopy(thisdict)

thisdict["nested"]["name"] = "mohamed"

print(thisdict)
print(mydict)
```

```python
# Iterating over dictionaries
# we can loop to iterate over dictionaries, keys, values, items.

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over values
for values in student.values():
    print(values)
```

```python
## Iterate over key value pairs
for key, value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested Disctionaries
students = {
    "student1": {"name": "Krish", "age": 32},
    "student2": {"name": "Peter", "age": 35},
}
print(students)
```

```python
## Access nested dictionaries elementss
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over nested dictionaries

for students_id, students_info in students.items():
    print(f"{students_id}: {students_info}")
    for key, value in students_info.items():
        print(f"{key}: {value}")
```

```python
## Dictionary Comphrehension

squares = {x: x**2 for x in range(5)}
print(squares)
```

```python
# Conditional Dictionary Comphrehension

evens = {x: x**2 for x in range(10) if x % 2 == 0}
print(evens)
```

```python
## Practical Examples
## USe a dictionary to count the frequency of elements in list
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```

```python
## Merge 2 dictionaries into one

dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
