Introduction of Tuples

Explanation
Tupules are ordered collection of items that are immutable. They are similar to lists, but their immutability makes them different

```python
empty_tuple = ()
print(empty_tuple)  # Output: ()
print(type(empty_tuple))  # Output: <class 'tuple'>
```

```python
lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
# converting tuples to list and list to tuples
numbers = tuple([1, 2, 3, 4, 5])
numbers
```

```python
list((1, 2, 3, 4, 5))
```

```python
mixed_tuple = (1, "Hello", 3.4, True)

print(mixed_tuple)
```

```python
# Accessing Tuple Elements

print(numbers[2])
print(numbers[-1])

print(numbers[::-1])
```

Tuple Operations

```python
concatenation_tuple = numbers + mixed_tuple

print(concatenation_tuple)
```

```python
mixed_tuple * 3 # Duplicate the tuple 3 times
```

```python
numbers * 3
```

```python
# Immutable nature of tuples
# Tuples are immutable, meaning their elements cannot be changed once assigned.

lst = [1, 2,3,4,5,6]
print(lst)

lst[1] = "ahmed"
print(lst)
```

```python
# Tuples are immutable, so you cannot change their elements after they are created. Attempting to do so will raise a TypeError.
numbers[1]='ahmed' # TypeError: 'tuple' object does not support item assignment
```

```python
numbers
```

```python
# Tuple Methods

print(numbers.count(1))
print(numbers.index(3)) # tuple.index(value) => Return first index of value.
```

```python
# Packing and Unpacking tuples

# Packing
packed_tuple = 1, "hello", 3.14
print(packed_tuple) #(1, 'hello', 3.14)

# UnPacking
a,b,c = packed_tuple
print(a)
print(b)
print(c)
```

```python
# unPacking with *

numbers = (1, 2, 3, 4, 5, 6)

first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

```python
# Nested Tuples

# Nested list
lst = [[1,2,3,4], [1,2,3,4], [1, "hello", 3.14, 'c']]
lst[0][0:3]
```

```python
lst = [[1,2,3,4], [1,2,3,4], (1, "hello", 3.14, 'c')]
lst[2][0:3]
```

```python
nested_tuple = ((1, 2, 3), ("a", "b", "c"), (True, False))

# Access the elements inside the tuple.

print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
# Iterating (Looping) over nested tuples

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
    print()
```

### Conclusion

Tuples are versatile and useful in many real-world scenarios where an immutable and ordered collection of items is required. They are commonly used in data structures, function arguments and return values, and as dictionary keys. Understanding how to leverage tuples effectively can improve the effeciency and readability of your python code.
